### **A simple code plotting a sinus function**

````python
#!/usr/bin/env python
import ROOT

print ("hello,␣world")
c = ROOT.TCanvas("can1", "Canvas␣title", 800, 600)
f = ROOT.TF1("fun1", "sin(x)",-5, 5)
f.Draw()

# draw canvas and save as a sin.png file
c.Draw()
c.SaveAs("sin.png")
````

## TRandom

Random number generator

````python
gg = []
for _ in range(10):
    gg.append(ROOT.gRandom.Gaus())
print([f"{x:.3f}" for x in gg])
````

And now filling histogram with a gaussian distribution

````python
import ROOT

h = ROOT.TH1F("h","Gaussian example",100,-5,5)

for _ in range(10000):
    h.Fill(ROOT.gRandom.Gaus())

c = ROOT.TCanvas("c1","",800,600)
h.Draw()
c.Draw()

````

## TH2F

two-dimensional histogram

````python
import ROOT

h_2dgaus = ROOT.TH2F("h_2dgaus", ";y;x;#entries", 100, -10, 10, 100, -10, 10);
for _ in range(10000):
    h_2dgaus.Fill(ROOT.gRandom.Gaus(), ROOT.gRandom.Gaus())

c = ROOT.TCanvas("c1","",800,600)
````

````python
h_2dgaus.Draw("SCAT")
c.Draw()
````

## Drawing multiple histograms

````python
h1 = ROOT.TH1F("h1", "h1 title", 100, -10, 10)
h2 = ROOT.TH1F("h2", "h2 title", 100, -10, 10)
h3 = ROOT.TH1F("h3", "h3 title", 100, -10, 10)

for _ in range(10000):
    h1.Fill(ROOT.gRandom.Gaus())
    h2.Fill(ROOT.gRandom.Gaus(1, 1))
    h3.Fill(ROOT.gRandom.Gaus(0, 2))
````

````python
h1.Draw()
h2.Draw("same")
h3.Draw("same")

c.Draw()
````

````python
h1.SetLineWidth(4)
h2.SetLineColor(2)
h3.SetLineStyle(3)
c.Draw()
````

## TLegend

````python
leg = ROOT.TLegend(0.2, 0.6, 0.35, 0.8)
leg.AddEntry(h1, "dist 1", "l")
leg.AddEntry(h2, "dist 2", "l")
leg.AddEntry(h3, "dist 3", "l")

leg.Draw()
c.Draw()
````

## SetOptStat

````python
h1.Draw()
ROOT.gStyle.SetOptStat(0)

c.Draw()
````

## TFile

write a histogram to the file and read it

````python
f_out = ROOT.TFile("file.root", "recreate")

h_abc = ROOT.TH1F("h_abc", "", 100, 0, 10)
h_abc.Fill(0.5)
h_abc.Fill(2.5)
f_out.Write()
f_out.Close()

f_in = ROOT.TFile("file.root")
hist = f_in.Get("h_abc")
hist.Draw()
c.Draw()
````

## Create TTree

````python
import ROOT
from array import array

f = ROOT.TFile("tree.root","RECREATE")

tree = ROOT.TTree("events","events")

x = array('f',[0.])
tree.Branch("x",x,"x/F")

for i in range(1000):
    x[0] = ROOT.gRandom.Gaus()
    tree.Fill()

tree.Write()
f.Close()
````

````python
f = ROOT.TFile("tree.root")

f.ls()

events = f.Get("events")
events.GetEntries()

events.Show(5)
````
